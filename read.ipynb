```python
import pyarrow as pa
import pandas as pd
from pathlib import Path




def transformLaps(year,place,quali):
  caminho2 = Path.cwd() / "silver" / "Data_Laps" /f"year={year}" /f"place={place}" /f"quali={quali}"
  arquivo2 = caminho2 / f'dados_voltas_{place}_{year}.parquet'
  df2 = pd.read_parquet(arquivo2,engine ='fastparquet')
  return df2
 

transformLaps(2025,'monaco','Q')
```
